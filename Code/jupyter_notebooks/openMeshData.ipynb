## find nearest coordinate code

```python
LATITUDE   LONGITUDE 
40.778980  -73.969250
40.701214  -74.009028
```

```python
import math

def find_nearest_coordinate(reference_point):
    """
    Finds the nearest coordinate in a set to a given reference point.

    Args:
        reference_point (tuple): A tuple (x, y) representing the reference point.
        coordinate_set (list): A list of tuples, where each tuple (x, y) 
                               represents a coordinate in the set.

    Returns:
        tuple: The nearest coordinate from the coordinate_set to the reference_point.
               Returns None if the coordinate_set is empty.
    """
    coordinate_set = [[40.778980, -73.969250],[40.701214, -74.009028]]
    nearest_coord = None
    min_distance = float('inf')  # Initialize with a very large number

    for coord in coordinate_set:
        # Calculate Euclidean distance
        distance = math.sqrt((reference_point[0] - coord[0])**2 + (reference_point[1] - coord[1])**2)

        if distance < min_distance:
            min_distance = distance
            nearest_coord = coord
    return nearest_coord
```
